```python
import numpy as np
from skimage.io import imread
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
```

```python
image = imread("screen_from_vid.jpg")[...,0].astype(np.uint8) / 255
plt.figure(figsize=(10, 10))
plt.imshow(image, cmap='gray')
plt.show()
```

```python
image.shape
```

```python
res_fft = np.fft.fft2(image)
res_fft_shift = np.fft.fftshift(res_fft)
res_fft_shift_log = np.log(np.abs(res_fft_shift))
```

```python
plt.figure(figsize=(10, 10))
plt.imshow(res_fft_shift_log, cmap='gray')
plt.show()
```

```python
fft_pca = PCA(n_components=1)
res_fft_pca = fft_pca.fit_transform(res_fft_shift_log)
```
