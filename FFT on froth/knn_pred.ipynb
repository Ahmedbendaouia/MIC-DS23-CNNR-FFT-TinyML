```python
import pickle
from functions import TenneurReel, predict_video, CalculMoyenneTenneursPredites
import pandas as pd
```

```python
mknn = pickle.load(open("./models/knn_model", 'rb'))
```

```python
test_videos = ["rl4_pb12-4.mp4", "rl1_pb16-4.mp4", "rl3_pb23-4.mp4", "rl1_pb9-5.mp4", "rl4_pb30-6.mp4"]
```

```python
final_results_df = pd.DataFrame(columns=["vidéo", "Cu_réel", "Cu_pred", "Fe_réel", "Fe_pred", "Pb_réel", "Pb_pred", "Zn_réel", "Zn_pred"])
```

```python
for test_vid in test_videos:
    TeunneurReel = TenneurReel(test_vid, "DataSource.xlsx")
    Teneur_predit = predict_video("../video_test/" + test_vid, mknn)
    Moyenne = CalculMoyenneTenneursPredites(Teneur_predit)
    final_results_df = pd.concat(
        [
            final_results_df, 
            pd.DataFrame([
                [test_vid, TeunneurReel[0], Moyenne[0], TeunneurReel[1], Moyenne[1], TeunneurReel[2], Moyenne[2], TeunneurReel[3], Moyenne[3]]
            ], 
            columns=["vidéo", "Cu_réel", "Cu_pred", "Fe_réel", "Fe_pred", "Pb_réel", "Pb_pred", "Zn_réel", "Zn_pred"])
        ], ignore_index=True)
    print(test_vid + " done")
```
