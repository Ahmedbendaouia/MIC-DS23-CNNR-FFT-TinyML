```python
import os
import numpy as np
from PIL import Image
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from xgboost import XGBClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score
from sklearn.decomposition import PCA
```

```python
train_path = "C:/Users/ychaf/Desktop/cat_vs_dog/train"
```

```python
img_size = (100, 100)
```

```python
features = []
labels = []
for file in os.listdir(train_path):
    if file.endswith(".jpg"):
        img = Image.open(os.path.join(train_path, file))
        img = img.resize(img_size)
        img = np.array(img)[...,0].astype(np.uint8) / 255
        res_fft = np.fft.fft2(img)
        res_fft_shift = np.fft.fftshift(res_fft)
        res_fft_log = np.log(abs(res_fft_shift))
        fft_pca = PCA(n_components=1)
        res_fft_pca = fft_pca.fit_transform(res_fft_log)
        res_fft_pca = res_fft_pca.reshape(-1)
        features.append(res_fft_pca)
        if file.startswith("cat"):
            labels.append(0)
        elif file.startswith("dog"):
            labels.append(1)
```

```python
features = np.array(features)
labels = np.array(labels)
```

```python
features.shape, labels.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.2, random_state=42)
```

SVM with FFT and PCA

```python
svm = SVC()
svm.fit(X_train, y_train)
y_pred = svm.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
f1_score(y_test, y_pred)
```

Logistic Regression with FFT and PCA

```python
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
f1_score(y_test, y_pred)
```

Random forest with FFT and PCA

```python
rf = RandomForestClassifier()
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
f1_score(y_test, y_pred)
```

XGBoost with FFT and PCA

```python
xgb = XGBClassifier()
xgb.fit(X_train, y_train)
y_pred = xgb.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
f1_score(y_test, y_pred)
```
