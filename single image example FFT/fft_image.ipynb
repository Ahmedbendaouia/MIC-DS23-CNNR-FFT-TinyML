```python
import numpy as np
from skimage.io import imread
import matplotlib.pyplot as plt  
```

```python
image = imread("image.jpg")[...,0].astype(np.uint8) / 255
```

```python
plt.imshow(image, cmap="gray")
plt.show()
```

```python
res_fft = np.fft.fft2(image)
```

```python
res_fft_shift = np.fft.fftshift(res_fft)
```

```python
res_fft_log = np.log(abs(res_fft_shift))
```

```python
plt.imshow(res_fft_log, cmap="gray")
plt.show()
```
