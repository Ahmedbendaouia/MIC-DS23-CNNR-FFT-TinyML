```python
import os
import numpy as np
from PIL import Image
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
path = "C:/Users/ychaf/Desktop/age_images"
```

```python
img_size = (100, 100)
```

```python
images = []
labels = []
for age in os.listdir(path):
    for i, file in enumerate(os.listdir(os.path.join(path, age))):
        if i == 40:
            break
        if file.endswith(".jpg"):
            img = Image.open(os.path.join(path, age, file))
            img = img.resize(img_size)
            img = np.array(img)[...,0].astype(np.uint8) / 255
            res_fft = np.fft.fft2(img)
            res_fft_shift = np.fft.fftshift(res_fft)
            res_fft_log = np.log(abs(res_fft_shift))
            img = res_fft_log.flatten()
            images.append(img)
            labels.append(int(age))
```

```python
images = np.array(images)
labels = np.array(labels)
```

```python
images.shape, labels.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(images, labels, test_size=0.2, random_state=42)
```

SVM

```python
svm = SVR()
svm.fit(X_train, y_train)
y_pred = svm.predict(X_test)
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred)
```

```python
r2_score(y_test, y_pred)
```

Linear Regression

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred)
```

```python
r2_score(y_test, y_pred)
```

XGBoost

```python
xgb = XGBRegressor()
xgb.fit(X_train, y_train)
y_pred = xgb.predict(X_test)
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred)
```

```python
r2_score(y_test, y_pred)
```
